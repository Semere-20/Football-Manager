# FM24 Moneyball Scouting — Lesson 1

**Goal:** Import your real FM24 scouting export, inspect its quality, and calculate one player's CB score using our original 60/30/10 model.

**Before you run:** Keep `Scouted_CB.html` in the **same folder** as this notebook. Open the notebook in Jupyter and run cells from the top, one at a time with **Shift + Enter**. No external data downloads are needed.

Our agreed categories are **Primary = 60%**, **Secondary = 30%**, **Tertiary = 10%**. We intentionally excluded Free Kick Taking, Long Shots, Long Throws, and Penalty Taking.

## Step 1 — Import Pandas and locate your export

- `pandas` is a Python library for working with tables.
- `Path` helps us refer to files without writing a hard-coded Windows or Mac folder path.
- `file_path.exists()` checks whether the file is actually where we expect.

```python
import pandas as pd
from pathlib import Path

file_path = Path("Scouted_CB.html")
print("Pandas version:", pd.__version__)
print("Export found:", file_path.exists())

assert file_path.exists(), "Put Scouted_CB.html in the same folder as the notebook."
```

## Step 2 — Read the Football Manager HTML table

`pd.read_html(...)` returns a **list** of HTML tables, even if there's only one. We select the first table with `[0]` and store it in a **DataFrame** named `df`.

`keep_default_na=False` prevents Pandas from automatically turning labels such as `N/A` into missing numeric values. `encoding="utf-8"` preserves player names and the `£` symbol correctly.

```python
df = pd.read_html(file_path, encoding="utf-8", keep_default_na=False)[0]

print("Players (rows):", len(df))
print("Columns:", len(df.columns))
display(df[["Name", "Club", "Pac", "Acc", "Hea", "Mar"]].head())
```

## Step 3 — Inspect column names and data issues

Our current export should show **80 rows** and **42 columns**. There are currently **69 player records** whose names are saved as `- -` and whose clubs are saved as `-`. These are genuinely absent in the exported HTML, not caused by Pandas.

We will **not invent player identities**. We can still evaluate the numbers, but we'll need the names before making a usable purchase shortlist.

```python
print("Column headers:")
print(df.columns.tolist())

print("\nMissing player names:", df["Name"].eq("- -").sum())
print("Missing clubs:", df["Club"].eq("-").sum())

display(df.loc[df["Name"].eq("- -"), ["Name", "Club", "Age", "Pac", "Acc"]].head())
```

## Step 4 — Define our own position model

Instead of copying anyone's scoring formula, we create three Python **lists** containing the exact FM24 abbreviations for our categories.

Important: in this export, `Nat` contains **Natural Fitness** (a numeric 1–20 attribute), not nationality. The `Rec` column is the scout's recommendation, not a player attribute.

```python
primary = [
    "Hea", "Mar", "Pas", "Tck", "Cmp", "Pos", "Jum", "Str",
    "Ant", "Cnt", "Dec", "Vis", "Pac", "Acc"
]

secondary = [
    "Agi", "Wor", "Fir", "Tec", "Agg", "Bra", "Dri", "Bal", "Sta"
]

tertiary = [
    "Det", "Tea", "Ldr", "OtB", "Nat", "Fla", "Cor", "Cro", "Fin"
]

required_attributes = primary + secondary + tertiary
missing_attributes = sorted(set(required_attributes) - set(df.columns))

print("Primary attributes:", len(primary))
print("Secondary attributes:", len(secondary))
print("Tertiary attributes:", len(tertiary))
print("Missing attributes:", missing_attributes)

assert not missing_attributes, "The export is missing a required attribute."
assert len(required_attributes) == len(set(required_attributes)), "Some attributes appear twice."
```

## Step 5 — Pick one player to practice on

Start with **Genar**, whose displayed attributes are all exact values. This keeps the first mathematics exercise simple. Other players have unknown scouting ranges, e.g. `7-10`, which we'll handle in a later lesson.

`str.startswith` searches for names beginning with `Genar`. `iloc[0]` selects the first matching player row.

```python
player = df.loc[df["Name"].str.startswith("Genar")].iloc[0]

display(player[["Name", "Club", "Hea", "Mar", "Pac", "Acc"]])
```

## Step 6 — Calculate the score

First, we calculate one **average** for each category. Then we combine the averages according to our chosen weights.

\[
\text{CB score} = 0.60\times P + 0.30\times S + 0.10\times T
\]

`.astype(float)` converts each attribute into a number (for this player, all are exact). `.mean()` calculates an average across that category's attributes.

**Check your understanding:** A 20 in one Primary attribute does *not* automatically add 12 points. It slightly raises the **Primary average**, which then contributes 60% of the final result.

```python
primary_average = player[primary].astype(float).mean()
secondary_average = player[secondary].astype(float).mean()
tertiary_average = player[tertiary].astype(float).mean()

cb_score = (
    primary_average * 0.60
    + secondary_average * 0.30
    + tertiary_average * 0.10
)

print(f"Player: {player['Name']}")
print(f"Primary average:   {primary_average:.2f}/20")
print(f"Secondary average: {secondary_average:.2f}/20")
print(f"Tertiary average:  {tertiary_average:.2f}/20")
print(f"Final CB score:    {cb_score:.2f}/20")
```
