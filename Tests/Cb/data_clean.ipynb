```python
from pathlib import Path # for working with file paths
#holds the path to the file we want to check
file_path = Path("../../src/Fully_Scouted_CB.html")

print(file_path.is_file())


print(Path.cwd())
```

```python
# Load our Football Manager scouting data
import pandas as pd

df = pd.read_html(
    file_path,
    encoding="utf-8",
    keep_default_na=False
)[0]

# Display the first five rows
df.head()
```

```python
#data cleaning 

#Show the original column names
before= df.columns
print("Before Cleaning:")
print("Number of columns:", len(before))
print(before)

#lets remove the unwanted columns such 'Inf' and 'Rec' as they dont provide any useful information for our analysis
df = df.drop(columns=['Inf', 'Rec'])

#check the remaining columns
after= df.columns
print("\nAfter Cleaning:")
print("Number of columns:", len(after))
print(after)
```

```python

# Check for missing values in our scouting data
missing_values = df.isna().sum()

# Display the number of missing values in each column
print("Missing Values in Each Column:")
print(missing_values)
```

```python
#check for empty strings in the dataframe
empty_strings = (df == '').sum()

print("\nEmpty Strings in Each Column:")
print(empty_strings.to_string())
```

```python

# Step 5 - Rename abbreviated attribute columns

# Store original column names for comparison
before = df.columns.tolist()

# Dictionary mapping old names to new names
column_names = {
    "Nat": "Natural Fitness",
    "Cro": "Crossing",
    "Det": "Determination",
    "Fin": "Finishing",
    "Fla": "Flair",
    "Ldr": "Leadership",
    "Pac": "Pace",
    "Sta": "Stamina",
    "Bal": "Balance",
    "Acc": "Acceleration",
    "Pas": "Passing",
    "Dri": "Dribbling",
    "Mar": "Marking",
    "Jum": "Jumping Reach",
    "Tec": "Technique",
    "Dec": "Decisions",
    "Cmp": "Composure",
    "Ant": "Anticipation",
    "Fir": "First Touch",
    "Hea": "Heading",
    "Bra": "Bravery",
    "Agi": "Agility",
    "Vis": "Vision",
    "Tea": "Teamwork",
    "Tck": "Tackling",
    "Str": "Strength",
    "Wor": "Work Rate",
    "Agg": "Aggression",
    "Cnt": "Concentration",
    "Pos": "Positioning"
}

# Rename the columns
df = df.rename(columns=column_names)

# Check the results
after = df.columns.tolist()

print("Before Renaming:")
print(before)

print("\nAfter Renaming:")
print(after)
```

```python

# Group Football Manager technical attributes
technical_attributes = [
    "Crossing",
    "Finishing",
    "Passing",
    "Dribbling",
    "Marking",
    "Technique",
    "First Touch",
    "Heading",
    "Tackling"
]

print(technical_attributes)
```

```python
#separate the name by splitting the 'Name' column into Name and nationality columns

#view date
print("Before:")
print(df[['Name']].head())

#split the 'Name' column into two new columns: 'Name' and 'Nationality' starting on the right side of the string and splitting on the last occurrence of " -"
df[['Name', 'Nationality']] = df['Name'].str.rsplit(
    " -", n=1, expand=True)

#View the changes made to the dataframe
print("\nAfter:")
print(df[["Name", "Nationality"]].head())

#now lets move nationality column to the second position in the dataframe
df.insert(1, 'Nationality', df.pop('Nationality'))

# Display the first five rows of the updated dataframe
print(df.columns.tolist())


```

```python
technical_df = df[["Name"] + ["Nationality"] + technical_attributes]

technical_df.head()
```

```python

# Set the output file path
output_Tech_df = file_path.with_name("Technical_Attributes.html")

# Create an HTML heading
title = "<h1>Technical Attributes</h1>"

# Convert the DataFrame into an HTML table
table = technical_df.to_html(index=False)

# Add CSS to center the heading and table
style = """
<style>
    h1 {
        text-align: center;
    }

    table {
        margin: 0 auto;
    }
</style>
"""
# Combine the title, style, and the table
html_content = title + style + table

# Save everything to an HTML file
output_Tech_df.write_text(html_content, encoding="utf-8")

# Confirm the file was created
print("File exists:", output_Tech_df.is_file())
```

```python
#save the technical attributes to html here
output_Tech_df = file_path.with_name("Technical_Attributes.html")

# Export the technical DataFrame to an HTML file
technical_df.to_html(output_Tech_df, index=False, encoding="utf-8")

# Confirm the file was created
print("Saved to:", output_Tech_df.resolve())
print("File exists:", output_Tech_df.is_file())
```
